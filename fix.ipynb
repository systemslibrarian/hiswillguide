```python

# Fix App.jsx — remove orphaned lines at end of file
path = '/workspaces/hiswillguide/src/App.jsx'
with open(path, 'r', encoding='utf-8') as f:
    lines = f.readlines()

print(f"Total lines before fix: {len(lines)}")
print("Last 10 lines:")
for i, line in enumerate(lines[-10:], start=len(lines)-9):
    print(f"  {i}: {repr(line)}")

# Truncate to 1011 lines
lines = lines[:1011]
with open(path, 'w', encoding='utf-8') as f:
    f.writelines(lines)

print(f"\nDone. File now has {len(lines)} lines.")
print("Last line:", repr(lines[-1]))
```
